```python
import ollama
response = ollama.chat(
    model = 'gemma2:2b',
    messages = [
        {'role': 'system', 'content': 'You are a helpful assistant'},
        {'role': 'user', 'content': 'What is Machine Learning'}    
    ]    
)

print(response['message']['content'])
```

```python
import ollama

stream = ollama.chat(
    model = 'gemma2:2b',
    messages = [
        {'role': 'system', 'content': 'You are a helpful assistant'},
        {'role': 'user', 'content': 'What is Machine Learning'},
        {'role': 'assistant', 'content': 'ML is a field of study that gives computers the ability to learn without being explicitly programmed'},
        {'role': 'user', 'content': 'Give me some examples of Machine Learning applications in short sentences'}    
    ],
    stream = True  
)

for chunk in stream:
    print(chunk['message']['content'], end='', flush=True)
```
