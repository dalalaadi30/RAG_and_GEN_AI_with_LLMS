```python
from groq import Groq

client = Groq()

chat_completion = client.chat.completions.create(
    model = 'llama-3.1-8b-instant',
    messages = [
        {'role': 'system', 'content': 'You are a helpful AI assitant'},
        {'role': 'user', 'content': 'What is Deep Learning?'}
    ]
)

print(chat_completion.choices[0].message.content)
```

```python
from groq import Groq

client = Groq()

chat_completion = client.chat.completions.create(
    model = 'llama-3.1-8b-instant',
    messages = [
        {'role': 'system', 'content': 'You are a helpful AI assitant'},
        {'role': 'user', 'content': 'What is Deep Learning?'},
        {'role': 'assistant', 'content': 'Deep learning is a subset of machine learning that involves the use of artificial neural networks.'},
        {'role': 'user', 'content': 'Give me some application'}
    ]
)

print(chat_completion.choices[0].message.content)
```
